*Step 1- Understand the Concept**

Own words - Exception handling is used to handle errors in our program.

Exception handling prevents our program from crashing by giving meaningful messages to users and allowing the program to continue running.

Errors are two types -

1. Syntax Error - Mistakes in writing the program, such as missing : or incorrect spelling. It happens before execution.

2. Runtime Error - The syntax is correct, but an error happens while the program is running.

Common Built-in Exceptions

Exception                    Description

ZeroDivisionError             Raised when dividing by zero
ValueError                   Raised when a function gets a valid type but an invalid value
TypeError                    Raised when an operation is performed on incompatible data types
NameError                    Raised when a variable is not defined
IndexError                  Raised when a list index is out of range
KeyError                    Raised when a dictionary key is not found
AttributeError              Raised when an object does not have the requested attribute
FileNotFoundError            Raised when a file does not exist
ImportError                 Raised when something cannot be imported from a module
ModuleNotFoundError          Raised when a module is not found

**ZeroDivisionError**

When it occurs:

ZeroDivisionError occurs when we try to divide a number by zero.

Example:

10 / 0

Correct way to handle:

Use try and except ZeroDivisionError.

```python
# ZeroDivisionError

try:
    balance = 5000
    people = int(input("Enter number of people sharing the money: "))

    amount_per_person = balance / people

    print("Each person gets:", amount_per_person)

except ZeroDivisionError:
    print("Number of people cannot be zero.")
```

***ValueError***

When it occurs:

ValueError occurs when a function receives the correct type of value but the value itself is not valid.

Example:

int("abc")

The string "abc" cannot be converted into an integer.

Correct way to handle:

Use try and except ValueError.

```python
## 5) finally BLOCK

# ValueError

try:
    age = int(input("Enter your age: "))
    print("your age is:", age)

except ValueError:
    print("Please enter a valid number.")
```


***TypeError***

When it occurs:

TypeError occurs when we perform an operation between incompatible data types.

Example:

"10" + 5

A string and an integer cannot be added directly.

Correct way to handle:

Use try and except TypeError.

```python
# TypeError

try:
    result = 10 + "20"
    print(result)

except TypeError:
    print("Cannot add string and integer.")
```

***NameError***

When it occurs:

NameError occurs when we try to use a variable that has not been defined.

Example:

print(name)

if name was never created, Python gives NameError.

Correct way to handle:

Check the variable name and use try and except NameError when required.

```python
# NameError

try:
    name = "Aadharsh"
    print(nam)

except NameError:
    print("Variable name is incorrect.")
```

***IndexError***

When it occurs:

IndexError occurs when we try to access a list element using an index that does not exist.

Example:

fruits = ["Apple", "Banana", "Mango"]

fruits[5]

The list has indexes from 0 to 2.

Correct way to handle:

Use try and except IndexError.

```python
# IndexError

try:
    cars = ["Audi", "Benz", "Bmw"]
    print(cars[3])

except IndexError:
    print("Invalid list index.")
```

***KeyError***

When it occurs:

KeyError occurs when we try to access a dictionary using a key that does not exist.

Example:

marks = {"Math": 90, "Science": 85}

marks["Social"]

The key "Social" does not exist.

Correct way to handle:

Use try and except KeyError.

```python
        # KeyError

try:
    student = {
        "name": "Aadharsh",
        "age": 22
    }

    print(student["city"])

except KeyError:
    print("city information is not available.")
```

***AttributeError***

When it occurs:

AttributeError occurs when an object does not have the attribute or method that we are trying to use.

Example:

num = 100

num.upper()

The upper() method is not available for an integer.

Correct way to handle:

Use try and except AttributeError.

```python
# AttributeError

try:
    name = "Aadharsh"
    print(name.append("gowtham"))

except AttributeError:
    print("String does not have an append method.")
```

***FileNotFoundError***

When it occurs:

FileNotFoundError occurs when we try to open a file that does not exist.

Example:

open("students.txt", "r")

If students.txt does not exist, Python gives FileNotFoundError.

Correct way to handle:

Use try and except FileNotFoundError.

```python
# FileNotFoundError

try:
    file = open("movie_details.txt", "r")
    print(file.read())
    file.close()

except FileNotFoundError:
    print("Movie details file was not found.")
else:
    print("Movie details loaded successfully.")
finally:
    print("Thank you for using the movie booking system.")
```

***ImportError***

When it occurs:

ImportError occurs when we try to import something from a module but that thing does not exist in the module.

Example:

from math import square

The math module does not have a function called square.

Correct way to handle:

Use try and except ImportError.

```python
# ImportError

try:
    from math import food_delivery
    
except ImportError:
    print("The required food delivery service is unavailable.")
else:
    print("Food delivery service loaded successfully.")
finally:
    print("Food delivery application started.")
```

***ModuleNotFoundError***

When it occurs:

ModuleNotFoundError occurs when we try to import a module that is not installed or does not exist.

Example:

import mymodule

If mymodule is not available, Python gives ModuleNotFoundError.

Correct way to handle:

Use try and except ModuleNotFoundError.

```python
# ModuleNotFoundError

try:
    import hospital_database

except ModuleNotFoundError:
    print("Hospital database module is not installed.")
else:
    print("Hospital database connected successfully.")
finally:
    print("Hospital application initialization completed.")
```

*Step 3 – Using try, except, else and finally**

try:
The code which may produce an error is written inside try.

except:
It handles the error when an exception occurs.

else:
It executes when there is no exception.

finally:
It executes whether an exception occurs or not.

```python
# using try, except, else and finally

try:
    quantity = int(input("Enter the number of products: "))

    price = 500
    total = quantity * price

except ValueError:
    print("Please enter a valid number.")

else:
    print("order placed successfully.")
    print("Total amount:", total)
finally:
    print("Thank you for shopping with us.")
```

*tep 4 – Raise**

raise is used when we want to create an exception manually.

For example, if the user enters zero as the second number, we can manually raise ZeroDivisionError.

```python
# using raise

try:
    age = int(input("Enter your age: "))
    
    if age < 18:
        raise ValueError("You must be 18 or older to apply.")

    print("You are eligible to apply for a driving licence")

except ValueError as e:
    print("Application failed:",e)
```

*Step 5 – Assert**

assert is used to check whether a condition is true.

If the condition is true, the program continues.

If the condition is false, Python raises AssertionError.

Example:

assert age >= 18, "Age must be 18 or above."

```python
# Using assert

marks = 85

assert 0 <= marks <= 100, "Marks must be between 0 and 100."

print("Marks are valid.")
print("Student marks:", marks)
```

*Step 6 – Custom Exceptions**

Custom Exceptions are user-defined exceptions.

We create our own exception class by inheriting from Exception.

For example:

PasswordError

We can use PasswordError when the password does not satisfy our requirement.

```python
#Custom Exception
class InsufficientBalanceError(Exception):
    pass


balance = 5000

try:
    amount = int(input("Enter withdrawal amount: "))

    if amount > balance:
        raise InsufficientBalanceError(
            "Insufficient balance for this withdrawal."
        )

    balance -= amount

except InsufficientBalanceError as e:
    print("Transaction failed:", e)

else:
    print("Withdrawal successful.")
    print("Remaining balance:", balance)

finally:
    print("Thank you for using the ATM.")
```
