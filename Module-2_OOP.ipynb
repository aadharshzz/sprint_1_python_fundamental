## 1. Class

### What is a Class?
A **class** is a blueprint/template used to create objects.

### Why use it?
Classes group related **data (attributes)** and **behavior (methods)** together.

### Syntax
```python
class ClassName:
    # attributes and methods
    pass
```

````python
class Employee:
    name = "Aadharsh"
    designation = "AI&ML"

print(Employee.name)
print(Employee.designation)
````

### Output
```text
Aadharsh
AI&ML
```

**Key idea:** The class describes what a Student object can contain or do.


## 2. Object

An **object** is an instance of a class.

### Why use objects?
A class is the blueprint; objects are the actual entities created from that blueprint.


````python
class Employee:
    def display(self):
        print("This is a Employee object.")

Employee1 = Employee()
Employee2 = Employee()

Employee1.display()
Employee2.display()
````

### Output
```text
This is a Employee object.
This is a Employee object.
```


## 3. Constructor – `__init__()`

`__init__()` is a special method that runs automatically when an object is created.

### Why use it?
It initializes an object's starting data.


````python
class employee:
    def __init__(self, name, age):
        self.name = name
        self.age = age

employee = employee("Aadharsh", 22)

print(employee.name)
print(employee.age)
````

### Output
```text
Aadharsh
22
```

`self` refers to the **current object**.


## 4. Instance Variables

Instance variables belong to a **particular object**. Different objects can have different values.


````python
class employee:
    def __init__(self, name, phoneno):
        self.name = name
        self.phoneno = phoneno

employee1 = employee("Aadharsh", 234567)
employee2 = employee("vijay", 12345675)

print(employee1.name, employee1.phoneno)
print(employee2.name, employee2.phoneno)
````

### Output
```text
Aadharsh 234567
vijay 12345675
```

## 5. Class Variables

A class variable is shared by all objects of  class.

### When to use it?
it can be used when a value should be common to every instance.


````python
class employee:
    company = "stackly"

    def __init__(self, name):
        self.name = name

employee1 = employee("Aadharsh")
employee2 = employee("vijay")

print(employee1.name, employee1.company)
print(employee2.name, employee2.company)
````

### Output
```text
Aadharsh stackly
vijay stackly
```

## 6. Instance Methods

Instance methods operate on object-specific data and normally receive `self` as the first parameter.


````python
class employee:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display(self):
        print(f"Name: {self.name}")
        print(f"age: {self.age}")

employee = employee("Aadharsh", 22)
employee.display()
````

### Output
```text
Name: Aadharsh
age: 22
```


## 7. Static Methods

A static method does not need `self` or `cls`. It behaves like a utility function that logically belongs to the class.

Use `@staticmethod`.


````python
class Calculator:
    @staticmethod
    def add(a, b):
        return a + b

print(Calculator.add(10, 20))
````

### Output
```text
30
```
